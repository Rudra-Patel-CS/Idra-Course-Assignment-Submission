```python
import pandas as pd
import numpy as np
import os

# Check if file exists, else create sample data for demonstration
file_path = "/mnt/data/Day11_Messy_Company_Employee_Dataset.csv"
if os.path.exists(file_path):
    df = pd.read_csv(file_path)
else:
    print(f"Warning: {file_path} not found. Using sample data.")
    data = {
        "Employee_ID": [101, 102, 101, 103],
        "Department": ["ENGINEERING", "engineering", "Engineering", "Sales"],
        "Gender": ["Male", "MALE", "Male", "Female"],
        "Age": [30, np.nan, 30, 25]
    }
    df = pd.DataFrame(data)

df.head()
```

# Load Data

```python
for col in ["City", "Work_Mode", "Job_Title", "Joining_Date"]:
    if col in df.columns:
        if col == "Joining_Date":
            print(f"\n--- {col} Analysis ---")
            print(df[col].head())
            print(df[col].dtype)
        else:
            print(f"\n--- {col} Value Counts ---")
            print(df[col].value_counts(dropna=False))
    else:
        print(f"Column '{col}' not found in current DataFrame.")

print("\n--- Missing Values Summary ---")
print(df.isnull().sum().sort_values(ascending=False))
```

# Fill Nulls

```python
cols_to_fill_mode = ["Department", "Gender", "City", "Work_Mode"]
cols_to_fill_median = ["Age", "Annual_Salary", "Experience_Years", "Performance_Score"]
for col in cols_to_fill_mode:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mode()[0])
for col in cols_to_fill_median:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())
display(df.isnull().sum())
```

# Fix Text  

```python
cols_to_fix = ["Department", "Gender", "City", "Work_Mode"]
for col in cols_to_fix:
    if col in df.columns:
        df[col] = df[col].str.strip().str.title()
        print(f"\n{col} Value Counts:")
        print(df[col].value_counts())
```

# Set Types

```python
if "Joining_Date" in df.columns:
    df["Joining_Date"] = pd.to_datetime(df["Joining_Date"], errors="coerce")
for col in ["Age", "Annual_Salary", "Experience_Years", "Performance_Score"]:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")
display(df.dtypes)
display(df.isnull().sum())
```

# Clean Row

```python
df = df.drop_duplicates().copy()
for col in ["Employee_Name", "Job_Title"]:
    if col in df.columns:
        df.loc[:, col] = df[col].str.strip()
if "Department" in df.columns:
    df.loc[:, "Department"] = df["Department"].replace({"Engineering": "Engineering", "ENGINEERING": "Engineering", "engineering": "Engineering"})
if "Gender" in df.columns:
    df.loc[:, "Gender"] = df["Gender"].replace({"Female": "Female", "female": "Female", "Male": "Male", "MALE": "Male", "Other": "Other"})
if "City" in df.columns:
    df.loc[:, "City"] = df["City"].replace({"delhi": "Delhi", "DELHI": "Delhi", "Delhi": "Delhi"})
if "Work_Mode" in df.columns:
    df.loc[:, "Work_Mode"] = df["Work_Mode"].replace({"REMOTE": "Remote", "remote": "Remote", "Remote": "Remote", "Office": "Office", "Hybrid": "Hybrid"})
```

# Final Check

```python
df.info()
display(df.describe())
print(f"Shape: {df.shape}")
num_cols = ["Annual_Salary", "Age", "Experience_Years", "Performance_Score"]
for col in num_cols:
    if col in df.columns:
        print(f"Mean {col}: {df[col].mean()}")
```

```python
import csv
import os
from datetime import datetime, timedelta
```

```python
BOOKS_FILE = "books.csv"
MEMBERS_FILE = "members.csv"
TRANSACTIONS_FILE = "transactions.csv"

LOAN_DAYS = 14
FINE_PER_DAY = 5
```

```python
class Book:

    def __init__(self, book_id, title, author, category, year):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.category = category
        self.year = year
        self.available = True

    def display(self):
        status = "Available" if self.available else "Issued"

        print(
            f"{self.book_id:<8}"
            f"{self.title:<30}"
            f"{self.author:<22}"
            f"{self.category:<18}"
            f"{self.year:<8}"
            f"{status:<10}"
        )

    def __str__(self):
        return f"{self.book_id} - {self.title} by {self.author}"
```

```python
class Member:

    def __init__(self, member_id, name, phone, email):
        self.member_id = member_id
        self.name = name
        self.phone = phone
        self.email = email
        self.borrowed_books = []

    def display(self):
        print(
            f"{self.member_id:<10}"
            f"{self.name:<25}"
            f"{self.phone:<18}"
            f"{self.email:<30}"
        )

    def borrow_book(self, book_id):
        self.borrowed_books.append(book_id)

    def return_book(self, book_id):
        if book_id in self.borrowed_books:
            self.borrowed_books.remove(book_id)

    def __str__(self):
        return f"{self.member_id} - {self.name}"
```

```python
class Library:
    def __init__(self):
        self.books = []
        self.members = []
        self.transactions = []

        # Set of supported library categories
        self.categories = {
            "Fiction",
            "Science",
            "Technology",
            "History",
            "Biography",
            "Education",
            "Business",
            "Programming",
            "Other"
        }

        # NOTE:
        # load_data() is called from main_menu() after all Library
        # methods have been attached in their separate notebook cells.
```

```python
def initialize_files(self):
    for file in [BOOKS_FILE, MEMBERS_FILE, TRANSACTIONS_FILE]:
        if not os.path.exists(file):
            with open(file, 'w', newline='') as f:
                writer = csv.writer(f)
                if file == BOOKS_FILE:
                    writer.writerow(["Book_ID", "Title", "Author", "Category", "Year", "Available"])
                elif file == MEMBERS_FILE:
                    writer.writerow(["Member_ID", "Name", "Phone", "Email"])
                elif file == TRANSACTIONS_FILE:
                    writer.writerow(["Transaction_ID", "Book_ID", "Member_ID", "Issue_Date", "Due_Date", "Return_Date", "Fine", "Status"])

Library.initialize_files = initialize_files
```

```python
def save_books(self):
    with open(BOOKS_FILE, "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["Book_ID", "Title", "Author", "Category", "Year", "Available"])
        for book in self.books:
            writer.writerow([book.book_id, book.title, book.author, book.category, book.year, book.available])

Library.save_books = save_books
```

```python
def save_members(self):
    with open(MEMBERS_FILE, "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["Member_ID", "Name", "Phone", "Email"])
        for member in self.members:
            writer.writerow([member.member_id, member.name, member.phone, member.email])

Library.save_members = save_members
```

# Save the Transactions

```python
def save_transactions(self):
    with open(TRANSACTIONS_FILE, "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["Transaction_ID", "Book_ID", "Member_ID", "Issue_Date", "Due_Date", "Return_Date", "Fine", "Status"])
        for t in self.transactions:
            writer.writerow([t["Transaction_ID"], t["Book_ID"], t["Member_ID"], t["Issue_Date"], t["Due_Date"], t["Return_Date"], t["Fine"], t["Status"]])

Library.save_transactions = save_transactions
```

```python
def load_data(self):
    """Load books, members, and transactions from CSV files."""

    self.initialize_files()

    # Load books
    try:
        with open(BOOKS_FILE, "r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            for row in reader:
                book = Book(
                    row["Book_ID"],
                    row["Title"],
                    row["Author"],
                    row["Category"],
                    row["Year"]
                )
                book.available = row["Available"] == "True"
                self.books.append(book)

    except (FileNotFoundError, KeyError, csv.Error) as error:
        print(f"Warning: Could not load books: {error}")

    # Load members
    try:
        with open(MEMBERS_FILE, "r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            for row in reader:
                member = Member(
                    row["Member_ID"],
                    row["Name"],
                    row["Phone"],
                    row["Email"]
                )
                self.members.append(member)

    except (FileNotFoundError, KeyError, csv.Error) as error:
        print(f"Warning: Could not load members: {error}")

    # Load transactions
    try:
        with open(TRANSACTIONS_FILE, "r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            for row in reader:
                self.transactions.append(dict(row))

    except (FileNotFoundError, KeyError, csv.Error) as error:
        print(f"Warning: Could not load transactions: {error}")

    # Restore books currently borrowed by members
    for transaction in self.transactions:
        if transaction.get("Status") == "Issued":
            member = self.find_member(transaction.get("Member_ID", ""))
            if member:
                member.borrow_book(transaction.get("Book_ID", ""))

Library.load_data = load_data
```

# Find Book

```python
def find_book(self, book_id):
    for book in self.books:
        if book.book_id.lower() == book_id.lower(): return book
    return None
Library.find_book = find_book
```

# Find Member

```python
def find_member(self, member_id):
    for m in self.members:
        if m.member_id.lower() == member_id.lower(): return m
    return None
Library.find_member = find_member
```

# Add Book

```python
def add_book(self):
    print("\n========== ADD NEW BOOK ==========")

    book_id = input("Enter Book ID: ").strip()

    if not book_id:
        print("Book ID cannot be empty.")
        return

    if self.find_book(book_id):
        print("Book ID already exists.")
        return

    title = input("Enter Book Title: ").strip()
    author = input("Enter Author Name: ").strip()

    if not title or not author:
        print("Title and author cannot be empty.")
        return

    print("\nAvailable Categories:")
    category_list = sorted(self.categories)

    for index, category in enumerate(category_list, start=1):
        print(f"{index}. {category}")

    try:
        choice = int(input("Select category number: "))

        if choice < 1 or choice > len(category_list):
            print("Invalid category selection.")
            return

        category = category_list[choice - 1]

    except ValueError:
        print("Invalid input. Please enter a category number.")
        return

    try:
        year = int(input("Enter Publication Year: "))

        if year < 0 or year > datetime.now().year:
            print("Invalid publication year.")
            return

    except ValueError:
        print("Invalid year. Please enter a valid number.")
        return

    book = Book(
        book_id,
        title,
        author,
        category,
        year
    )

    self.books.append(book)
    self.save_books()

    print("\nBook added successfully.")
    print(f"Book ID  : {book.book_id}")
    print(f"Title    : {book.title}")
    print(f"Author   : {book.author}")
    print(f"Category : {book.category}")
    print(f"Year     : {book.year}")

Library.add_book = add_book
```

# View books

```python
def view_books(self):
    if not self.books: return print("Empty.")
    for b in self.books: b.display()
Library.view_books = view_books
```

# Search Books

```python
def search_books(self):
    print("\n========== SEARCH BOOK ==========")

    keyword = input(
        "Enter title, author, category, or Book ID: "
    ).strip().lower()

    if not keyword:
        print("Search term cannot be empty.")
        return

    found = False

    print("\nSearch Results:")
    print("-" * 96)

    for book in self.books:
        if (
            keyword in book.book_id.lower()
            or keyword in book.title.lower()
            or keyword in book.author.lower()
            or keyword in book.category.lower()
        ):
            book.display()
            found = True

    if not found:
        print("No matching books found.")

Library.search_books = search_books
```

# Add Member

```python
def add_member(self):
    print("\n========== ADD MEMBER ==========")

    member_id = input("Enter Member ID: ").strip()

    if not member_id:
        print("Member ID cannot be empty.")
        return

    if self.find_member(member_id):
        print("Member ID already exists.")
        return

    name = input("Enter Member Name: ").strip()
    phone = input("Enter Phone Number: ").strip()
    email = input("Enter Email: ").strip()

    if not name or not phone or not email:
        print("All member details are required.")
        return

    member = Member(
        member_id,
        name,
        phone,
        email
    )

    self.members.append(member)
    self.save_members()

    print("\nMember added successfully.")

Library.add_member = add_member
```

# View Member

```python
def view_members(self):
    for m in self.members: m.display()
Library.view_members = view_members
```

# Issue Book

```python
def issue_book(self):
    print("\n========== ISSUE BOOK ==========")

    book_id = input("Enter Book ID: ").strip()
    member_id = input("Enter Member ID: ").strip()

    if not book_id or not member_id:
        print("Book ID and Member ID are required.")
        return

    book = self.find_book(book_id)
    member = self.find_member(member_id)

    if not book:
        print("Book not found.")
        return

    if not member:
        print("Member not found.")
        return

    if not book.available:
        print("This book is already issued.")
        return

    if len(member.borrowed_books) >= 3:
        print("A member can borrow a maximum of 3 books.")
        return

    issue_date = datetime.now()
    due_date = issue_date + timedelta(days=LOAN_DAYS)

    transaction_number = len(self.transactions) + 1
    transaction_id = f"T{transaction_number:04d}"

    transaction = {
        "Transaction_ID": transaction_id,
        "Book_ID": book.book_id,
        "Member_ID": member.member_id,
        "Issue_Date": issue_date.strftime("%Y-%m-%d"),
        "Due_Date": due_date.strftime("%Y-%m-%d"),
        "Return_Date": "",
        "Fine": "0",
        "Status": "Issued"
    }

    self.transactions.append(transaction)
    book.available = False
    member.borrow_book(book.book_id)

    self.save_books()
    self.save_transactions()

    print("\nBook issued successfully.")
    print(f"Transaction ID : {transaction_id}")
    print(f"Book           : {book.title}")
    print(f"Member         : {member.name}")
    print(f"Issue Date     : {issue_date.strftime('%Y-%m-%d')}")
    print(f"Due Date       : {due_date.strftime('%Y-%m-%d')}")

Library.issue_book = issue_book
```

# Return Book

```python
def return_book(self):
    print("\n========== RETURN BOOK ==========")

    book_id = input("Enter Book ID: ").strip()

    if not book_id:
        print("Book ID cannot be empty.")
        return

    book = self.find_book(book_id)

    if not book:
        print("Book not found.")
        return

    transaction = None

    for record in self.transactions:
        if (
            record.get("Book_ID", "").lower() == book_id.lower()
            and record.get("Status") == "Issued"
        ):
            transaction = record
            break

    if not transaction:
        print("This book is not currently issued.")
        return

    today = datetime.now()
    due_date = datetime.strptime(
        transaction["Due_Date"],
        "%Y-%m-%d"
    )

    late_days = max((today - due_date).days, 0)
    fine = late_days * FINE_PER_DAY

    transaction["Return_Date"] = today.strftime("%Y-%m-%d")
    transaction["Fine"] = str(fine)
    transaction["Status"] = "Returned"

    member = self.find_member(transaction["Member_ID"])

    if member:
        member.return_book(book_id)

    book.available = True

    self.save_books()
    self.save_transactions()

    print("\nBook returned successfully.")
    print(f"Book       : {book.title}")
    print(f"Return Date: {today.strftime('%Y-%m-%d')}")
    print(f"Late Days  : {late_days}")
    print(f"Fine       : ₹{fine}")

Library.return_book = return_book
```

# Currently issued book

```python
def issued_books(self):
    print("\n========== CURRENTLY ISSUED BOOKS ==========")

    issued = [
        transaction
        for transaction in self.transactions
        if transaction.get("Status") == "Issued"
    ]

    if not issued:
        print("No books are currently issued.")
        return

    print(
        f"{'Transaction':<15}"
        f"{'Book ID':<12}"
        f"{'Member ID':<12}"
        f"{'Issue Date':<15}"
        f"{'Due Date':<15}"
    )

    print("-" * 69)

    for transaction in issued:
        print(
            f"{transaction['Transaction_ID']:<15}"
            f"{transaction['Book_ID']:<12}"
            f"{transaction['Member_ID']:<12}"
            f"{transaction['Issue_Date']:<15}"
            f"{transaction['Due_Date']:<15}"
        )

Library.issued_books = issued_books
```

# Overdue Books

```python
def overdue_books(self):
    print("\n========== OVERDUE BOOKS ==========")

    today = datetime.now().date()
    found = False

    for transaction in self.transactions:
        if transaction.get("Status") != "Issued":
            continue

        due_date = datetime.strptime(
            transaction["Due_Date"],
            "%Y-%m-%d"
        ).date()

        if today > due_date:
            late_days = (today - due_date).days
            estimated_fine = late_days * FINE_PER_DAY

            print(f"\nTransaction : {transaction['Transaction_ID']}")
            print(f"Book ID     : {transaction['Book_ID']}")
            print(f"Member ID   : {transaction['Member_ID']}")
            print(f"Due Date    : {transaction['Due_Date']}")
            print(f"Late Days   : {late_days}")
            print(f"Estimated Fine: ₹{estimated_fine}")

            found = True

    if not found:
        print("No overdue books.")

Library.overdue_books = overdue_books
```

# Remove book

```python
def remove_book(self):
    print("\n========== REMOVE BOOK ==========")

    book_id = input("Enter Book ID: ").strip()
    book = self.find_book(book_id)

    if not book:
        print("Book not found.")
        return

    if not book.available:
        print("Cannot remove an issued book.")
        return

    confirmation = input(
        f"Remove '{book.title}'? (y/n): "
    ).strip().lower()

    if confirmation == "y":
        self.books.remove(book)
        self.save_books()
        print("Book removed successfully.")
    else:
        print("Operation cancelled.")

Library.remove_book = remove_book
```

# Remove Member

```python
def remove_member(self):
    print("\n========== REMOVE MEMBER ==========")

    member_id = input("Enter Member ID: ").strip()
    member = self.find_member(member_id)

    if not member:
        print("Member not found.")
        return

    if member.borrowed_books:
        print(
            "Cannot remove this member because they have borrowed books."
        )
        return

    confirmation = input(
        f"Remove member '{member.name}'? (y/n): "
    ).strip().lower()

    if confirmation == "y":
        self.members.remove(member)
        self.save_members()
        print("Member removed successfully.")
    else:
        print("Operation cancelled.")

Library.remove_member = remove_member
```

```python
def statistics(self):
    total_books = len(self.books)

    available_books = sum(
        1 for book in self.books
        if book.available
    )

    issued_books = total_books - available_books
    total_members = len(self.members)
    total_transactions = len(self.transactions)

    total_fines = sum(
        float(transaction.get("Fine", 0) or 0)
        for transaction in self.transactions
    )

    print("\n========================================")
    print("          LIBRARY STATISTICS")
    print("========================================")
    print(f"Total Books           : {total_books}")
    print(f"Available Books       : {available_books}")
    print(f"Issued Books          : {issued_books}")
    print(f"Registered Members    : {total_members}")
    print(f"Total Transactions    : {total_transactions}")
    print(f"Total Fines Collected : ₹{total_fines:.2f}")
    print("========================================")

Library.statistics = statistics
```

# Main Start Point..

```python
def main_menu():

    # Create the Library object only after all Library method cells
    # have been executed.
    library = Library()

    # Load saved CSV data after all methods have been attached.
    library.load_data()

    while True:
        print("\n")
        print("=" * 55)
        print("             LIBRARY MANAGEMENT SYSTEM")
        print("=" * 55)

        print("1.  Add New Book")
        print("2.  View All Books")
        print("3.  Search Book")
        print("4.  Remove Book")
        print("-" * 55)
        print("5.  Add New Member")
        print("6.  View Members")
        print("7.  Remove Member")
        print("-" * 55)
        print("8.  Issue Book")
        print("9.  Return Book")
        print("10. View Issued Books")
        print("11. View Overdue Books")
        print("-" * 55)
        print("12. Library Statistics")
        print("13. Exit")
        print("=" * 55)

        choice = input("Enter your choice: ").strip()

        try:
            if choice == "1":
                library.add_book()
            elif choice == "2":
                library.view_books()
            elif choice == "3":
                library.search_books()
            elif choice == "4":
                library.remove_book()
            elif choice == "5":
                library.add_member()
            elif choice == "6":
                library.view_members()
            elif choice == "7":
                library.remove_member()
            elif choice == "8":
                library.issue_book()
            elif choice == "9":
                library.return_book()
            elif choice == "10":
                library.issued_books()
            elif choice == "11":
                library.overdue_books()
            elif choice == "12":
                library.statistics()
            elif choice == "13":
                print("\nThank you for using the Library Management System.")
                print("Have a great day!")
                break
            else:
                print("\nInvalid choice!")
                print("Please select an option from 1 to 13.")
        except Exception as error:
            print("\nAn unexpected error occurred.")
            print(f"Error: {error}")
```

```python
# Validate the Library setup before starting the application
required_methods = [
    "initialize_files",
    "save_books",
    "save_members",
    "save_transactions",
    "load_data",
    "find_book",
    "find_member",
    "add_book",
    "view_books",
    "search_books",
    "remove_book",
    "add_member",
    "view_members",
    "remove_member",
    "issue_book",
    "return_book",
    "issued_books",
    "overdue_books",
    "statistics"
]

missing_methods = [
    method for method in required_methods
    if not hasattr(Library, method)
]

if missing_methods:
    raise RuntimeError(
        "Some Library methods are missing. Run all cells from the beginning: "
        + ", ".join(missing_methods)
    )

print("Library setup validated successfully.")
print("All required methods are available.")
```

```python
# Start the Library Management System
main_menu()
```

# 40. Passenger Load Factor

This is a useful **data transformation** and makes your analysis more meaningful.

### Formula

```text
Passenger Load Factor =
Passengers / Seat Capacity × 100
```

````python
df["Load_Factor"] = (df["Passengers"] / df["Seat_Capacity"]) * 100
display(df[["Flight_ID", "Passengers", "Seat_Capacity", "Load_Factor"]].head())
````

# 41. Average Load Factor

````python
print("Average Load Factor:", round(df["Load_Factor"].mean(), 2), "%")
````

# 42. Find Highly Utilized Flights

````python
high_load = df[df["Load_Factor"] > 90]
display(high_load[["Flight_ID", "Airline", "Passengers", "Seat_Capacity", "Load_Factor"]].head(10))
````

# 43. Revenue Estimate

We can create another useful transformed column.

### Formula

```text
Estimated Revenue =
Passengers × Average Ticket Price
```

````python
df["Estimated_Revenue"] = (df["Passengers"] * df["Average_Ticket_Price"])
display(df[["Flight_ID", "Passengers", "Average_Ticket_Price", "Estimated_Revenue"]].head())
````

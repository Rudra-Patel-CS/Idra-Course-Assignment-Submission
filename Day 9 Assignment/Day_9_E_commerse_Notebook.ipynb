# Processed E-commerce Dataset

In this notebook, I will process three e-commerce datasets: Orders, Customers, and Products.

I will use Pandas to load, merge, combine, transform, and export the data into one clean dataset.

```python
import pandas as pd
```

# Load Orders Dataset

### Load Datasets

```python
orders = pd.read_csv('Day9_Orders.csv')
print('Orders dataset loaded successfully!')
```

# Load Customers Dataset

```python
customers = pd.read_csv('Day9_Customers.csv')
print('Customers dataset loaded successfully!')
```

# Load Products Dataset

```python
products = pd.read_csv('Day9_Products.csv')
print('Products dataset loaded successfully!')
```

# Explore Datasets

### Explore Datasets

```python
display(orders.head())
display(customers.head())
display(products.head())
```

```python
print('Orders:', orders.shape)
print('Customers:', customers.shape)
print('Products:', products.shape)
```

# Using `merge()`

### Merge Datasets

```python
orders_customers = pd.merge(orders, customers, on='Customer_ID', how='left')
display(orders_customers.head())
```

```python
processed_data = pd.merge(orders_customers, products, on='Product_ID', how='left')
display(processed_data.head())
```

# Using `concat()`

### Demonstrate Concatenation

```python
orders_part1 = orders.iloc[:60]
orders_part2 = orders.iloc[60:]
combined_orders = pd.concat([orders_part1, orders_part2], ignore_index=True)
print('Original Orders:', len(orders))
print('After concat:', len(combined_orders))
```

# DateTime Operations

### DateTime Transformations

```python
processed_data['Order_Date'] = pd.to_datetime(processed_data['Order_Date'])
processed_data['Order_Year'] = processed_data['Order_Date'].dt.year
processed_data['Order_Month'] = processed_data['Order_Date'].dt.month
processed_data['Order_Day'] = processed_data['Order_Date'].dt.day
processed_data['Day_of_Week'] = processed_data['Order_Date'].dt.day_name()
display(processed_data[['Order_Date', 'Order_Year', 'Order_Month', 'Day_of_Week']].head())
```

# Using `apply()`

### Calculate Sales and Categories

```python
processed_data['Total_Sales'] = processed_data.apply(lambda row: row['Quantity'] * row['Unit_Price'], axis=1)
processed_data['Order_Size'] = processed_data['Total_Sales'].apply(lambda x: 'High' if x >= 5000 else 'Normal')
display(processed_data[['Product_Name', 'Quantity', 'Unit_Price', 'Total_Sales', 'Order_Size']].head())
```

# Create a Clean Dataset

### Final Dataset Selection

```python
cols = ['Order_ID', 'Order_Date', 'Order_Year', 'Order_Month', 'Order_Day', 'Day_of_Week', 'Customer_ID', 'Customer_Name', 'City', 'Region', 'Membership_Type', 'Product_ID', 'Product_Name', 'Category', 'Brand', 'Quantity', 'Unit_Price', 'Total_Sales', 'Payment_Method', 'Order_Status', 'Order_Size']
final_data = processed_data[cols]
display(final_data.head())
```

### Basic Analysis and Export

```python
print('Total Sales:', final_data['Total_Sales'].sum())
display(final_data.groupby('Category')['Total_Sales'].sum())
```

# Export the Processed Dataset

```python
final_data.to_csv('Day9_Processed_Ecommerce_Dataset.csv', index=False)
print('Processed dataset exported successfully!')
```

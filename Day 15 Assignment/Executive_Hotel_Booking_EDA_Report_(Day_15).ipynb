# 6. Data Cleaning

The dataset contains missing values in fields such as `Children`, `Agent_ID`, and `Company_ID`.

For this analysis, we don't want to unnecessarily delete thousands of records. We'll handle them sensibly.

```python
# Fill missing children values with 0

df["Children"] = df["Children"].fillna(0)
```

```python
# Fill missing Agent and Company IDs with 0

df["Agent_ID"] = df["Agent_ID"].fillna(0)
df["Company_ID"] = df["Company_ID"].fillna(0)
```

```python
# Check missing values again

df.isnull().sum().sum()
```

# 33. Outlier Detection Using IQR

We'll use a simple IQR method for important numerical columns.

```python
# Function to find outliers using IQR

def find_outliers(column):

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    return outliers
```

```python
# Check ADR outliers

adr_outliers = find_outliers("ADR")

print(
    "Number of ADR Outliers:",
    len(adr_outliers)
)
```

```python
# Check lead time outliers

lead_time_outliers = find_outliers(
    "Lead_Time_Days"
)

print(
    "Number of Lead Time Outliers:",
    len(lead_time_outliers)
)
```

```python
# Check estimated revenue outliers

revenue_outliers = find_outliers(
    "Estimated_Revenue"
)

print(
    "Number of Revenue Outliers:",
    len(revenue_outliers)
)
```

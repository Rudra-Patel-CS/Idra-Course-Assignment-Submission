```python
import random
```

# Set The Limit

```python
LOWER_LIMIT = 1
UPPER_LIMIT = 100
MAX_ATTEMPTS = 7
```

```python
# Generate a random number
secret_number = random.randint(LOWER_LIMIT, UPPER_LIMIT)

print("===================================")
print("     NUMBER GUESSING GAME")
print("===================================")
print(f"Guess the number between {LOWER_LIMIT} and {UPPER_LIMIT}")
print(f"You have {MAX_ATTEMPTS} attempts.")
print("Good Luck!\n")
```

# MAIN LOGIC FOR GUESSSING THE NUMBER

```python

attempt = 1

while attempt <= MAX_ATTEMPTS:
    guess = int(input(f"Attempt {attempt}: Enter your guess: "))

    if guess < secret_number:
        print("Too Low! Try Again.\n")
    elif guess > secret_number:
        print("Too High! Try Again.\n")
    else:
        print(f"\n Congratulations! You guessed the correct number in {attempt} attempt(s).")
        break

    attempt += 1
```

## Number is not guessed And Attempts Are Finish..

```python

if attempt > MAX_ATTEMPTS:
    print("\nGame Over!")
    print(f"The correct number was: {secret_number}")
```

```python

print("\n===================================")
print("Thank you for playing!")
print("===================================")
```

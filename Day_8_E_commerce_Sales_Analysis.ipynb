## Sort by Highest Sales

```python
highest_sales = df.sort_values(by="Total_Sales", ascending=False)
highest_sales.head(10)
```

## Sort by Quantity

```python
df.sort_values(by="Quantity", ascending=False).head(10)
```

## Total Sales Statistics

```python
total_sales = df["Total_Sales"].sum()
print("Total Sales:", total_sales)
```

```python
average_sales = df["Total_Sales"].mean()
print("Average Sales:", average_sales)
```

```python
highest_sale = df["Total_Sales"].max()
print("Highest Sale:", highest_sale)
```

```python
lowest_sale = df["Total_Sales"].min()
print("Lowest Sale:", lowest_sale)
```

```python
total_quantity = df["Quantity"].sum()
print("Total Quantity Sold:", total_quantity)
```

```python
number_of_orders = df["Order_ID"].nunique()
print("Number of Orders:", number_of_orders)
```

## GroupBy Analysis

```python
category_sales = df.groupby("Category")["Total_Sales"].sum()
print(category_sales)
```

```python
category_sales.sort_values(ascending=False)
```

```python
category_quantity = df.groupby("Category")["Quantity"].sum()
print(category_quantity)
```

```python
city_sales = df.groupby("City")["Total_Sales"].sum()
print(city_sales)
```

```python
city_sales.sort_values(ascending=False).head(5)
```

```python
product_sales = df.groupby("Product")["Total_Sales"].sum()
print(product_sales)
```

```python
product_sales.sort_values(ascending=False).head(5)
```

```python
payment_sales = df.groupby("Payment_Method")["Total_Sales"].sum()
print(payment_sales)
```

```python
payment_sales.sort_values(ascending=False)
```

## Aggregation Summary

```python
sales_summary = df["Total_Sales"].agg(["sum", "mean", "max", "min"])
print(sales_summary)
```

```python
category_summary = df.groupby("Category")["Total_Sales"].agg(["sum", "mean", "max", "min"])
category_summary
```

## Top Performing Orders

```python
top_orders = df.sort_values("Total_Sales", ascending=False).head(10)
top_orders[["Order_ID", "Product", "Category", "City", "Total_Sales"]].head(10)
```

```python
top_product = product_sales.idxmax()
top_product_sales = product_sales.max()
print("Top Product:", top_product)
print("Total Sales:", top_product_sales)
```

```python
top_category = category_sales.idxmax()
top_category_sales = category_sales.max()
print("Top Category:", top_category)
print("Total Sales:", top_category_sales)
```

## Overall Summary

```python
print("========== E-COMMERCE SALES SUMMARY ==========")
print(f"Total Orders       : {number_of_orders}")
print(f"Total Quantity Sold: {total_quantity}")
print(f"Total Sales        : {total_sales:.2f}")
print(f"Average Sales      : {average_sales:.2f}")
print(f"Highest Sale       : {highest_sale:.2f}")
print(f"Lowest Sale        : {lowest_sale:.2f}")
print(f"Top Category       : {top_category}")
print(f"Top Product        : {top_product}")
```

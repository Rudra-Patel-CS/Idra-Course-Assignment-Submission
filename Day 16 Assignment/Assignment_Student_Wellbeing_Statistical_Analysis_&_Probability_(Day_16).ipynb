# 13. IQR Outlier Detection


```python
# Function to find outliers using IQR

def find_outliers(column):

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    return outliers, lower_limit, upper_limit
```

```python
# Check outliers in all required variables

outlier_variables = [
    "Weekly_Study_Hours",
    "Daily_Screen_Time_Hours",
    "Commute_Time_Minutes",
    "Monthly_Discretionary_Spending"
]

for column in outlier_variables:

    outliers, lower, upper = find_outliers(column)

    print(column)
    print("Lower Limit:", round(lower, 2))
    print("Upper Limit:", round(upper, 2))
    print("Number of Outliers:", len(outliers))
    print()
```

# 23. P(A or B)

```python
# Probability of A or B

P_A_or_B = (A | B).mean()

print("P(A or B) =", round(P_A_or_B, 4))
```

# 24. P(A and B)

```python
# Probability of A and B

P_A_and_B = (A & B).mean()

print("P(A and B) =", round(P_A_and_B, 4))
```

# 25. P(A|B)

```python
# Probability of A given B

P_A_given_B = (A & B).sum() / B.sum()

print("P(A|B) =", round(P_A_given_B, 4))
```

# 26. P(B|A)

```python
# Probability of B given A

P_B_given_A = (A & B).sum() / A.sum()

print("P(B|A) =", round(P_B_given_A, 4))
```

# 27. Are Year 1 and Year 4 Mutually Exclusive?

# 28. Test Independence of A and B

```python
# Check independence

product = P_A * P_B

print("P(A and B):", round(P_A_and_B, 4))
print("P(A) × P(B):", round(product, 4))
```

# 29. Bayes' Theorem

```python
# Calculate P(B|not A)

not_A = ~A

P_B_given_not_A = (B & not_A).sum() / not_A.sum()

print(
    "P(B|not A) =",
    round(P_B_given_not_A, 4)
)
```

# 30. Calculate P(A|B) Using Bayes' Theorem

### Code Block 35

```python
# Bayes' theorem

P_A_bayes = (
    P_B_given_A * P_A
) / (
    P_B_given_A * P_A
    + P_B_given_not_A * (1 - P_A)
)

print(
    "P(A|B) using Bayes' theorem =",
    round(P_A_bayes, 4)
)
```

---

# 31. Verify Bayes' Theorem

```python
# Compare direct and Bayes results

print(
    "Direct P(A|B):",
    round(P_A_given_B, 4)
)

print(
    "Bayes P(A|B):",
    round(P_A_bayes, 4)
)

print(
    "Difference:",
    round(abs(P_A_given_B - P_A_bayes), 10)
)
```

# 33. Mean and Standard Deviation

### Code Block 37

```python
# Mean and standard deviation

readiness = df["Academic_Readiness_Score"]

mean_readiness = readiness.mean()
std_readiness = readiness.std()

print("Mean:", round(mean_readiness, 2))
print("Standard Deviation:", round(std_readiness, 2))
```

# 34. Find Highest and Lowest Scores

```python
# Find highest and lowest academic readiness scores

highest_score = readiness.max()
lowest_score = readiness.min()

print("Highest Score:", highest_score)
print("Lowest Score:", lowest_score)
```

---

# 35. Z-score of Highest Score

```python
# Z-score of highest score

z_highest = (
    highest_score - mean_readiness
) / std_readiness

print(
    "Z-score of highest score:",
    round(z_highest, 2)
)
```

# 36. Z-score of Lowest Score

```python
# Z-score of lowest score

z_lowest = (
    lowest_score - mean_readiness
) / std_readiness

print(
    "Z-score of lowest score:",
    round(z_lowest, 2)
)
```

# 37. 68-95-99.7 Empirical Rule
